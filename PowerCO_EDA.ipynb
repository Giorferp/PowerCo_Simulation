### Churn exploration

```python
# pyright: reportUndefinedVariable=false, reportMissingTypeStubs=false, reportAttributeAccessIssue=false, reportArgumentType=false
churn = client_data[['id', 'churn']]
total = len(churn)

plt.figure(figsize=(6, 5))
ax = sns.countplot(
    data=churn, x='churn', hue='churn', palette='magma'
)

    for c in ax.containers:
        if hasattr(c, "datavalues"):  # type: ignore
            labels = [f"{(v / total) * 100:.1f}%" if v > 0 else "" for v in getattr(c, "datavalues", [])]
            ax.bar_label(
                c,  # type: ignore
      labels=labels,
      label_type='center',
      color='white',
      fontsize=12,
      weight='bold',
  )

plt.xticks([0, 1], ['Retenido (0)', 'Churn (1)'])
plt.ylabel('Cantidad de Clientes')
plt.grid(axis='y', alpha=0.25)
plt.title('Churn Distribution', weight='bold')
plt.tight_layout()
plt.show()
```

### Electricity and gas usage

```python
# pyright: reportUndefinedVariable=false, reportMissingTypeStubs=false, reportAttributeAccessIssue=false, reportArgumentType=false
client_data_log = pd.DataFrame(
    {
        'cons_12m_log': np.log10(client_data['cons_12m'] + 1),
        'cons_gas_12m_log': np.log10(client_data['cons_gas_12m'] + 1),
    }
)

plt.figure(figsize=(10, 5))
sns.boxplot(data=client_data_log, palette='Set2')
plt.ylabel('Electric usage')
plt.title('Usage distribution (Logaritmic Scale)')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
